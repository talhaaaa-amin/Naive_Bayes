```python
import math     
```

```python
class NaiveBayes:
    """A simple class to Implement NaiveBayes.""" # Docstrings explain what the class does

    def __init__(self):
        print(self)

    def Prior(self,total_emails)->list:
        N:int = sum(i for i in total_emails) # get the n    umber of total emails
        print(f'Total emails are {N}')
        return [i/N for i in total_emails]
        
    def likeLihood(self,feature_counts,total_emails,new_emails)->list:
        lik = []
        for i in range(0,len(feature_counts)):
            V = len(feature_counts[i])  
            lik.append([
                ( (word + 1) / (total_emails[i] + V) ) ** new_emails[w]
                  for w,word in enumerate(feature_counts[i])
                ])
            
        return lik
    
    def posterior(self,prior,likelihood):
        arr = []
        for i in range(0,len(prior)):
            arr.append(prior[i]* math.prod(likelihood[i]))
        return arr

    def Predictor(self,total_emails,feature_counts,new_emails):
        PriorList = self.Prior(total_emails)
        print(f'Prior is : {PriorList}')
        like = self.likeLihood(feature_counts,total_emails,new_emails=new_emails)

        #Now using the origional Naive bayes we are going to predice the Possibility of Spam and Not Spam
        post= self.posterior(PriorList,like)
        print(f'likelihood is {like} \nPosterior is:{post}')
        
        if(post[0] > post[1]):
            return ("Email is Spam")
        return  "Email is not Spam"
   
```

```python
def main():
    # Total emails of each class
    total_emails = [50, 70]  # [Spam, Not Spam]

    # Number of emails containing each word per class
    # Order of features: [Free, Win, Offer]
    feature_counts = [
        [40, 35, 45],  # Spam
        [10, 15, 20]   # Not Spam
    ]

    # Features in the new email (1 = present, 0 = absent)
    new_email = [1, 1, 1]  # Free, Win, Offer

    n1 = NaiveBayes()
    out = n1.Predictor(total_emails=total_emails,feature_counts=feature_counts,new_emails=new_email)
    print(out)
    return 

if __name__ == "__main__":
    main()
```
