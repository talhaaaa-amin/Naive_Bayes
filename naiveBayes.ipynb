```python
class NaiveBayes:
    """A simple class to Implement NaiveBayes."""  # Docstrings explain what the class does

    def __init__(self):
        print(self)

    def Prior(self, total_emails) -> list:
        N: int = sum(i for i in total_emails)  # get the n    umber of total emails
        print(f"Total emails are {N}")
        return [i / N for i in total_emails]

    def likeLihood(self, feature_counts, total_emails, new_emails) -> list:
        lik = []
        for i in range(0, len(feature_counts)):
            V = len(feature_counts[i]) #here V is the vocabulary
            lik.append(
                [
                    ((word + 1) / (total_emails[i] + V)) ** new_emails[w]
                    for w, word in enumerate(feature_counts[i])
                ]
            )

        return lik

    def posterior(self, prior, likelihood):
        arr = []
        for i in range(0, len(prior)):
            arr.append(prior[i] * math.prod(likelihood[i]))
        return arr

    def Predictor(self, total_emails, feature_counts, new_emails):
        PriorList = self.Prior(total_emails)
        print(f"Prior is : {PriorList}")
        like = self.likeLihood(feature_counts, total_emails, new_emails=new_emails)

        # Now using the origional Naive bayes we are going to predice the Possibility of Spam and Not Spam
        post = self.posterior(PriorList, like)
        print(f"likelihood is {like} \nPosterior is:{post}")

        if post[0] > post[1]:
            return "Email is Spam"
        return "Email is not Spam"
```

```python
class Bernoulli_Naive_Bayes:
    """A simple class to Implement Bernoulli NaiveBayes."""  # Docstrings explain what the class does

    def __init__(self):
        print(self)

    def Prior(self, total_emails) -> list:
        N: int = sum(i for i in total_emails)  # get the number of total emails
        print(f"Total emails are {N}")
        return [math.log(i / N) for i in total_emails]

    def likeLihood(
        self, feature_counts, total_emails, new_emails
    ) -> list:  # Added laplace smoothing
        lik = []
        for i in range(0, len(feature_counts)):
            V = len(feature_counts[i])
            lik.append(
                [
                    math.log(((word + 1) / (total_emails[i] + V))) * new_emails[w]
                    + math.log(1 - ((word + 1) / (total_emails[i] + V)))
                    * (
                        1 - new_emails[w]
                    )  # numerator +1 and denominator + V number of words in Spam class
                    for w, word in enumerate(feature_counts[i])
                ]
            )

        return lik

    def posterior(self, prior, likelihood):
        arr = []
        for i in range(0, len(prior)):
            arr.append(prior[i] + sum(likelihood[i]))
        return arr

    def Predictor(self, total_emails, feature_counts, new_emails):
        PriorList = self.Prior(total_emails)
        print(f"Prior is : {PriorList}")
        like = self.likeLihood(feature_counts, total_emails, new_emails=new_emails)

        # Now using the origional Naive bayes we are going to predice the Possibility of Spam and Not Spam
        post = self.posterior(PriorList, like)
        print(f"likelihood is {like} \nPosterior is:{post}")

        if post[0] > post[1]:
            return "Email is Spam"
        return "Email is not Spam"
```

```python
class Bernoulli_Naive_Bayes_quiz:
    def __init__(self):
        self.documents = []
        self.labels = []
        self.class_priors = {}
        self.feature_probs = {}

    def train(self, documents:list, labels:list):
        # calculate:
        # 1. class priors

        self.documents = documents
        # Convention: 1 for positive, 0 for negative
        self.labels = [1 if label == "positive" else 0 for label in labels]
        
        P = sum(self.labels)  # positive count
        N = len(self.labels) - P  # negative count

        # Fixed keys to match the logic (0 for negative, 1 for positive)
        self.class_priors[1] = P / (P + N)  
        self.class_priors[0] = N / (P + N)  

        print(f"class Priors are: {self.class_priors}")

        # Lets fetch the vocabulary from the data and remove punctuation
        tokenized_docs = []
        vocabulary = []

        for doc in self.documents:
            words = []
            for raw_word in doc.lower().split():
                clean_word = "".join(ch for ch in raw_word if ch.isalpha())
                if clean_word:
                    words.append(clean_word)
                    if clean_word not in vocabulary:
                        vocabulary.append(clean_word)
            tokenized_docs.append(words) # Fixed indentation

        # Build Bernoulli features (1 if word exists in document, else 0)
        X_train = []
        for words in tokenized_docs:
            word_set = set(words)
            X_train.append([1 if v in word_set else 0 for v in vocabulary])

        y_train = self.labels

        self.vocabulary = vocabulary
        self.X_train = X_train
        self.y_train = y_train

        print("vocabulary =", self.vocabulary)
        print("X_train =", self.X_train)
        print("y_train =", self.y_train)

        # 2. Compute feature probabilities P(x_i=1 | class) with Laplace smoothing
        for c in [0, 1]:
            # Get indices of documents belonging to class c
            class_indices = [idx for idx, label in enumerate(y_train) if label == c]
            N_c = len(class_indices)  # number of docs in class c

            probs = []
            for j in range(len(vocabulary)):
                # Count how many docs in class c have feature j = 1
                count = sum(X_train[idx][j] for idx in class_indices)
                # Laplace smoothing: (count + 1) / (N_c + 2)
                probs.append((count + 1) / (N_c + 2))
            self.feature_probs[c] = probs

        print("feature_probs =", self.feature_probs)

    def predict(self, X_test):
        # apply Bernoulli Naive Bayes formula
        predictions = []
        for x in X_test:
            class_scores = {}
            for c in [0, 1]:
                # Start with log prior
                log_prob = math.log(self.class_priors[c])
                # Add log-likelihood for each feature using Bernoulli formula
                for j in range(len(self.vocabulary)):
                    p = self.feature_probs[c][j]  # P(x_j=1 | c)
                    if x[j] == 1:
                        log_prob += math.log(p)
                    else:
                        log_prob += math.log(1 - p)
                class_scores[c] = log_prob

            # Predict the class with the highest log-posterior
            predicted_class = max(class_scores, key=class_scores.get)
            predictions.append("positive" if predicted_class == 1 else "negative")

        print("Predictions:", predictions)
        return predictions
```

```python
def main():
    documents = [
        "Amazing movie with great acting and excellent story",
        "I loved this film, it was wonderful and inspiring",
        "Great direction and good music, really enjoyed it",
        "Terrible movie, boring plot and bad acting",
        "Waste of time, very dull and disappointing film",
    ]

    labels = ["positive", "positive", "positive", "negative", "negative"]

    model = Bernoulli_Naive_Bayes_quiz()
    model.train(documents=documents, labels=labels)

    # Build test vectors using the learned vocabulary
    test_docs = [
        "great acting in this film",
        "terrible boring and bad movie",
    ]

    X_test = []
    for doc in test_docs:
        words = set(doc.lower().split())
        X_test.append([1 if v in words else 0 for v in model.vocabulary])

    print("\nTest documents:", test_docs)
    print("X_test =", X_test)
    model.predict(X_test)


if __name__ == "__main__":
    main()
```
